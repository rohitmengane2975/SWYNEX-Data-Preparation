```python
import pandas as pd

df = pd.read_csv("train.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
# Age: fill missing values with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Embarked: fill missing values with mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Cabin: drop because most values are missing
df = df.drop(columns=['Cabin'])

# Check missing values again
print(df.isnull().sum())
```

```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Check data types
print(df.dtypes)
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nDataset shape:", df.shape)
```

```python
df.to_csv("SWYNEX_Titanic_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## Data Preparation Assumptions

1. Missing Age values were replaced using the median Age.

2. Missing Embarked values were replaced using the mode.

3. The Cabin column was removed because a large proportion
   of its values were missing.

4. Duplicate rows were checked and no duplicate rows were found.

5. Data types were reviewed and found appropriate.
